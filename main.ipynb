```python
import dataset_loader
import display
import data_analyzer
import importlib
# importlib.reload(dataset_loader)
# importlib.reload(display)
# importlib.reload(data_analyzer)

from dataset_loader import get_file_names, create_appliance_objects
from display import suggest_reduction
from data_analyzer import DataAnalyzer

data_files = get_file_names()   # get the list of files in the /data directory

appliance_dict = create_appliance_objects(data_files)   # get a dictionary with all of the appliances

data_analyzer = DataAnalyzer()   # instantiate all the data analyzer

for key in appliance_dict:
    data_analyzer.add_appliance(appliance_dict[key])    # go through and add all of the appliance objects to the DataAnalyzer class

total_time_dict = data_analyzer.get_total_time()    # get a dictionary containing all of the available metrics from the DataAnalyzer class
rms_power_dict = data_analyzer.get_rms_power()
energy_cost_dict = data_analyzer.get_energy_cost()

appliance_list = list(appliance_dict.values())    # get a list of all the appliances for the display functions
suggest_reduction(appliance_list)
```

```python
# CLI Tool to get more metrics about each of the appliances
while True:
    print("Available appliances:")
    for name in appliance_dict.keys():
        print(f"- {name}")
    print("- quit (to exit)")

    appliance = input("Enter the name of the appliance (or 'quit' to exit): ").strip()
    if appliance.lower() == 'quit':
        break
    while appliance not in appliance_dict:
        print("Invalid appliance name. Please choose from the list above or 'quit' to exit.")
        appliance = input("Enter the name of the appliance (or 'quit' to exit): ").strip()
        if appliance.lower() == 'quit':
            break
    if appliance.lower() == 'quit':
        break

    print("\nAvailable metrics:")
    print("1 - total_time")
    print("2 - rms_power")
    print("3 - energy_cost")

    metric = input("Enter the number of the metric you want: ").strip()
    if metric.lower() == 'quit':
        break
    while metric not in ["1", "2", "3"]:
        print("Invalid number. Please choose from: 1, 2, 3")
        metric = input("Enter the number of the metric you want: ").strip()
        if metric.lower() == 'quit':
            break
    if metric.lower() == 'quit':
        break

    if metric == "1":
        print(f"\nTotal time for {appliance}: {total_time_dict[appliance]:.2f}")
    elif metric == "2":
        print(f"\nRMS power for {appliance}: {rms_power_dict[appliance]:.2f} W")
    else:  # energy_cost
        kwh, cost = energy_cost_dict[appliance]
        print(f"\nEnergy cost for {appliance}: {kwh:.2f} kWh, ${cost:.2f}")
    print()  # Add a blank line for readability
```
